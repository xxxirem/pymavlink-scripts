```python
img = cv.imread("aruco_board.png")

print(img.shape)
```

Initializing Aruco

```python
ARUCO_DICT = cv.aruco.DICT_6X6_50
dict_ = cv.aruco.getPredefinedDictionary(ARUCO_DICT)
params = cv.aruco.DetectorParameters()
detector = cv.aruco.ArucoDetector(dict_, params)
```

```python
corners, ids, rejected = detector.detectMarkers(img)
print(corners)
print(rejected)
```

```python
print(ids)
for num in ids:
    print(num)
print(ids.flatten())
```

```python
import matplotlib.pyplot as plt
cp_img = img.copy()
cv.aruco.drawDetectedMarkers(cp_img, corners, ids)
plt.figure(figsize=(8, 8))
plt.imshow(cp_img)
plt.axis("off")
plt.show()
```
